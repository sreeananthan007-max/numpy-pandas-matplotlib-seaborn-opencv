```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("stud.csv")
df
```

```python
%matplotlib inline
plt.xlabel("Study Hours (X)")
plt.xlabel("Exam Marks (Y)")
plt.scatter(df["Study Hours (X)"],df["Exam Marks (Y)"],color='red',marker='o')

```

```python
ndf=df.drop("Exam Marks (Y)",axis="columns")
ndf
```

```python
y=df["Exam Marks (Y)"]
y
```

```python
reg=linear_model.LinearRegression()
reg.fit(ndf,y)
```

```python
reg.predict([[12]])
```

```python
reg.coef_
```
